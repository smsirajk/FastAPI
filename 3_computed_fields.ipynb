##Computed_field - which is calculated using the existing field in the dataset

```python
from pydantic import computed_field, Field, BaseModel, EmailStr, AnyUrl, field_validator
from typing import List, Dict, Optional, Annotated

class Patient(BaseModel):
    name: str
    email: EmailStr
    age: int
    weight: float
    height: float # computed field
    married: bool
    allergies: List[str] = None
    contact_details: Dict[str, str]

    @computed_field
    @property
    def calculate_bmi(self) -> float:
        bmi = self.weight / (self.height ** 2)
        return bmi
    
def update_patient_data(pat_info: Patient):
    print(pat_info.name)
    print(pat_info.email)
    print(pat_info.age)
    print(pat_info.weight) 
    print(pat_info.weight)
    print('BMI :',pat_info.calculate_bmi)
    print(pat_info.allergies)
    print(pat_info.married)
    print('Updated...')


patient_info = {'name':'Mustafa','email':'abc@hdfc.com', 'age': '62', 'weight':75.2,'height':1.67, 'married':True, 'contact_details':{'email':'abc@gmail.com', 'phone':'123145154', 'emergency':'191'}}

patient1 = Patient(**patient_info)

update_patient_data(patient1)
    
```
