#3 How to Restrict the convertion, some cases, pydantic smartly change the datatype string to interger, as example - weight ='65.5', it will smartly convert and return to 65.5

How would we restrict it? 



```python
from pydantic import BaseModel, Field
from typing import List, Dict, Optional, Annotated

class patient(BaseModel):
     name: Annotated[str, Field(max_length =50, title='Name of the Patient', description='Give the name of the patient less than 50', examples=['Mieshaa', 'Minnat'])]
     age: int = Field (gt=0, lt = 120)
     weight: Annotated[float, Field(gt=0, strict=True)]
     married: Annotated[bool, Field(default=None, description='Is the patient married or not')] # default value set
     alergies: Annotated[Optional[List[str]], Field(default=None, max_length=5)] #Default value pass
     contact_details: Dict[str, str]

def insert_patient_data(pat_info: patient):
    print(pat_info.name)
    print(pat_info.age)
    print(pat_info.weight) 
    print(pat_info.alergies)
    print(pat_info.married)
    print('Inserted')


patient_info = {'name':'Mustafa', 'age': 111, 'weight':75.2, 'married':True, 'contact_details':{'email':'abc@gmail.com', 'phone':'123145154'}}

patient1 = patient(**patient_info)

insert_patient_data(patient1)
```
