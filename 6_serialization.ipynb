```python
from pydantic import BaseModel

class Adress(BaseModel):
    city: str
    state: str
    pin: str

class  Patient(BaseModel):
    name: str
    gender: str
    age: int 
    address: Adress

address_dict = {'city': 'Kolkata', 'state': 'West Bengal', 'pin':'700161'}
address1 = Adress(**address_dict)

patient_dict = {'name':'Mieshaa', 'gender': 'Female', 'age': 8, 'address':address1}

patient1 = Patient(**patient_dict)

print(patient1)
print(patient1.name)
print(patient1.address.city)
print(patient1.address.state)

temp = patient1.model_dump(exclude={'address':['state']})

temp1 = patient1.model_dump(include={'address':['state']})

print(temp)

print(temp1)

print(type(temp))
```

```python
from pydantic import BaseModel

class Adress(BaseModel):
    city: str
    state: str
    pin: str

class  Patient(BaseModel):
    name: str
    gender: str = 'Male'
    age: int 
    address: Adress

address_dict = {'city': 'Kolkata', 'state': 'West Bengal', 'pin':'700161'}
address1 = Adress(**address_dict)

patient_dict = {'name':'Mieshaa', 'age': 8, 'address':address1}

patient1 = Patient(**patient_dict)

print(patient1)
print(patient1.name)
print(patient1.address.city)
print(patient1.address.state)

temp = patient1.model_dump_json(exclude={'address':['state']})

temp1 = patient1.model_dump_json(exclude_unset=True)
print(temp)

print(temp1)

print(type(temp))
```
