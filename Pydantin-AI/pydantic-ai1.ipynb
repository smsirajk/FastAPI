what is Pydantic?
Pydantic is a library that makes sure your data is exactly what you  expect it to be. 

In standard Python, we can add 'type hints' to suggest what data type a variable should be. But Python doesn't actually st rict enforce it. Let's look at an example.

```python
from pydantic import BaseModel  

class game(BaseModel):
    score: int
    name: str
    team: str

def play_game(score: int):
    print("Welcome to the game!")
    # game logic here

    print(f"Your score is :{score}")

# We said score is an integer, but we have to pass a string to the function. 
# Python doesn't care it just runs it anyway, but Pydantic will throw an error if the type is not correct.

play_game(100)
```

```python
from pydantic import BaseModel, Field, field_validator


class UserSignUp(BaseModel):
    username: str = Field(min_length=3, max_length=15) # Must be between 3 and 15 characters
    age : int = Field(ge = 13) # Must be greated than or equal
    account_type: str = "free" # Default value if nothing is given doesn't even need field 

# We ca nwrite completely custom rules useing @field_validator decorator.
@field_validator('username')
@classmethod
def check_username_no_spaces(cls, value: str) -> str:
    if  " " in value:
        raise ValueError("Username cannot contain spaces")
    return value

@field_validator('username')
@classmethod
def check_username_no_special_characters(cls, value: str) -> str:
    if " " in value or not value.isalnum():
        raise ValueError("Username cannot contain special characters")
    return value


```
